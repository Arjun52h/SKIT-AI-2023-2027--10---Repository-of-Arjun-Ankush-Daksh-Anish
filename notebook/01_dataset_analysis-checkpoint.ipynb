```python
df.groupby("event")["visitorid"].nunique() # tell us how many unique user have the interaction of view , transcation etc 
```

```python
top_products = (
    df.groupby("itemid").size()
    .sort_values(ascending = False).head(10)
)

top_products # give top 10 product having more interaction in whole dataset
```

```python
sample_user = df["visitorid"].iloc[2]

sample_user
```

```python
df[df["visitorid"] == sample_user].head(20)
```

```python
top_product_events = (
    df[df["itemid"].isin(top_products.index)]
    .groupby(["itemid", "event"])
    .size()
    .unstack(fill_value=0)
)
# tell which top 10 prdct have interactions higher 
top_product_events
```

```python
user_interactions = df.groupby("visitorid").size() # means each user has how much interactions , jitni baar user aaya utni interactions 
user_interactions.sort_values(ascending=False).head(10)
```

```python
df["datetime"] = pd.to_datetime(df["timestamp"], unit="ms")

print("Start Date:", df["datetime"].min())
print("End Date:", df["datetime"].max())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
user_activity = df.groupby("visitorid").size()

print("Minimum interactions:", user_activity.min())
print("Maximum interactions:", user_activity.max())
print("Average interactions:", user_activity.mean())
print("Median interactions:", user_activity.median())
```

```python
product_activity = df.groupby("itemid").size()

print("Minimum interactions:", product_activity.min())
print("Maximum interactions:", product_activity.max())
print("Average interactions:", product_activity.mean())
print("Median interactions:", product_activity.median())
```
